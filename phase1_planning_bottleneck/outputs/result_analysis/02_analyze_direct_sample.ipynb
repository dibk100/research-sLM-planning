기본 입력 파일: `outputs/direct_stdin/results.jsonl`

```python
from pathlib import Path
import json

import matplotlib.pyplot as plt
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', 120)
```

```python
RESULT_PATH = Path('../direct_stdin/results.jsonl')

if not RESULT_PATH.exists():
    raise FileNotFoundError(
        f'Result file not found: {RESULT_PATH.resolve()}'
    )

records = []
with RESULT_PATH.open('r', encoding='utf-8') as file:
    for line_number, line in enumerate(file, start=1):
        if not line.strip():
            continue
        try:
            records.append(json.loads(line))
        except json.JSONDecodeError as error:
            raise ValueError(f'Invalid JSON at line {line_number}') from error

df = pd.DataFrame(records)
print(f'Loaded records: {len(df)}')
print(f'Columns: {len(df.columns)}')
# display(df.head())
```

한 개의 샘플(첫 번째 문제)을 사람이 읽기 쉽게 뜯어보기

```python
import json

sample = df.iloc[0]         # 첫 번째 문제

print("=" * 100)
print(
    f"{sample['problem_id']} | "
    f"{sample['title']} | "
    f"{sample['difficulty']} | "
    f"{sample['status']}"
)
print("=" * 100)

print("\n[Dataset Information]")
print(f"Platform     : {sample['platform']}")
print(f"Contest ID   : {sample['contest_id']}")
print(f"Contest Date : {sample['contest_date']}")

print("\n[Problem]")
print(sample["problem"][:1000])

print("\n[Formatted Prompt]")
print(sample["formatted_prompt"][:1000])

print("\n[Generation Statistics]")
print(f"Prompt tokens     : {sample['prompt_tokens']}")
print(f"Completion tokens : {sample['completion_tokens']}")
print(f"Generation time   : {sample['generation_time']:.4f}s")

print("\n[Evaluation Summary]")
print(f"Passed            : {sample['passed']}")
print(f"Status            : {sample['status']}")
print(f"Passed tests      : {sample['passed_tests']} / {sample['total_tests']}")
print(f"Execution time    : {sample['execution_time']:.4f}s")

print("\n[Error message]")
print(sample["error_message"])

print("\n[Raw Output]")
print(sample["raw_output"])

print("\n[Extracted Code]")
print(sample["extracted_code"])

print("\n[All Test Results]")
print(
    json.dumps(
        sample["test_results"],
        indent=2,
        ensure_ascii=False,
    )
)
```

첫 번째 실패한 테스트만

```python
failed_test = next(
    (
        test
        for test in sample["test_results"]
        if not test["passed"]
    ),
    None,
)

if failed_test is not None:
    print("=" * 80)
    print("First Failed Test")
    print("=" * 80)

    print(f"Test index : {failed_test['test_index']}")
    print(f"Status     : {failed_test['status']}")

    print("\n[Input]")
    print(failed_test["input_text"])

    print("\n[Expected]")
    print(failed_test["expected_output"])

    print("\n[Actual]")
    print(failed_test["actual_output"])

    print("\n[stderr]")
    print(failed_test["stderr"])
```
